
# Analiza konvergencije i kriterijumi zaustavljanja optimizacionih algoritama

## Nasumična pretraga (Random Search)

Analiza konvergencije:
Metode nasumične pretrage za diskretnu stohastičku optimizaciju generišu nizove rešenja koji obično formiraju **Markovljeve lance**. Pod specifičnim uslovima (povezanost i konačnost prostora stanja), ovi algoritmi konvergiraju **skoro sigurno** (a.s.) ili u verovatnoći ka skupu globalnih optimalnih rešenja. Teorijska stopa konvergencije za procenu optimalne vrednosti funkcije kod mnogih varijanti iznosi $m^{-1/2}$, gde je $m$ broj iteracija.

Kriterijumi zaustavljanja:
U teorijskim razmatranjima, kriterijum zaustavljanja se često izostavlja jer je konvergencija asimptotska. U praksi, pretraga se prekida nakon dostizanja **unapred definisanog budžeta evaluacija** ili **maksimalnog broja iteracija**.

## Koordinatni spust (Coordinate Descent) sa Zlatnim presekom

Analiza konvergencije:
Koordinatni spust (CD) pokazuje **linearnu konvergenciju** na separabilnim funkcijama. Korišćenje **Zlatnog preseka** (Fibonacci metoda) unutar koraka spusta nudi stopu konvergencije $c \approx -0.4812/d$, gde je $d$ dimenzionalnost problema. Adaptivne varijante, poput ACiD (Adaptive Coordinate Descent), koriste adaptivno kodiranje (AE) kako bi transformisale koordinatni sistem, čime se drastično ubrzava konvergencija na neseparabilnim funkcijama (npr. Rosenbrock) pretvarajući ih u "prividno separabilne".

Kriterijumi zaustavljanja:
Algoritam se zaustavlja kada poboljšanje najbolje vrednosti funkcije postane zanemarljivo (npr. manje od $10^{-25}$) tokom određenog broja evaluacija, ili kada se dostigne \maksimalni budžet evaluacija (tipično $10^4 \cdot d$).

## Nelder-Mead

Analiza konvergencije:
Konvergencija se analizira putem **beskonačnih proizvoda matrica transformacije**. Za funkcije u dve dimenzije koje su strogo konveksne, dokazano je da **prečnik simpleksa teži nuli**, a vrednosti funkcije u temenima konvergiraju ka istoj vrednosti. U višim dimenzijama (dokazano do 8D), konvergencija ka tački je garantovana ako koraci kontrakcije dominiraju. Međutim, algoritam može zakazati konvergirajući ka nestaćionarnoj tački u specifičnim slučajevima.

Kriterijumi zaustavljanja:
Primarni kriterijumi su tolerancija vrednosti funkcije (tolfun) i tolerancija promene koordinata temena (tolx).

## CMA-ES (Covariance Matrix Adaptation Evolution Strategy)

Analiza konvergencije:
CMA-ES postiže **log-linearnu konvergenciju** pri skoro optimalnim stopama. Algoritam je invarijantan na rotaciju i translaciju, što omogućava identične performanse na čitavim klasama funkcija. Ključ uspeha je **adaptacija matrice kovarijanse**, koja uči inverznu Hessianovu matricu kod kvadratnih funkcija, transformišući problem u sfernu funkciju.

Kriterijumi zaustavljanja:
CMA-ES koristi skup sofisticiranih uslova:

*    tolfun: Razlika vrednosti funkcije ispod praga.
*    tolx: Promena parametara rešenja ispod praga.
*    Stagnacija: Nema poboljšanja kroz određeni broj generacija ($100 + 100n^{1.5}/\text{popsize}$).
*    tolconditioncov: Prekid ako matrica kovarijanse postane loše uslovljena ($> 10^{14}$).
*    Divergencija: Ako se korak previše poveća (tolfacupx).
